```python
class Food:
    category = "Foodies"

```

```python
# creating object
apple1 = Food()
apple1.category = "Super electronics"
blueberry1 = Food()
banana = Food()

print(apple1.category)
print(blueberry1.category) 
print(banana.category)
```

```python

class Phone:
    category = "Electronics"

    # Constructor
    def __init__(self, model, battery, camera):
        self.model = model
        self.battery = battery
        self.camera = camera


# Creating objects
apple = Phone("iPhone 15", 3000, 40)
blueberry = Phone("B15", 4500, 20)
banana = Phone("S25", 5000, 100)

print(apple.camera)
```
