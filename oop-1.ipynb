```python
# parent / base class 
class Phone:
    category = "Electronics"

    # Constructor
    def __init__(self, model, battery, camera, battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    # methods 
    def __init__(self, hour):
        self.charge += hour

    def capture(self, total_photo):
        if(self.battery_percentage) <= 0:
            print("No Charge")
        else:
            self.battery_percentage -= total_photo
            print(f"Photo captured in {self.model}")

    def charge(self, hour):
        print(f"Charging for {hour} hour(s)")


# Creating objects
apple = Phone("iPhone 15", 3000, 40)
blueberry = Phone("B15", 4500, 20)
banana = Phone("S25", 5000, 100)

apple.capture(20)

print(apple.camera)
print(blueberry.camera)
print(blueberry.battery_percentage)
print(apple.battery_percentage)
```

```python
# child / derived class

class smartphone(Phone):
    def __init__(self, model, battery, camera, processor):
        super().__init__(model, battery, 48, processor)
        self.processor = processor

    def charge(self, hour):
        print("fast charging in process")
        super().charge(hour)

oneplus = smartphone("Oneplus Nord", 4190, 100, "Snapdragon")    

oneplus.charge(1)

print(oneplus.model)
print(oneplus.charge)
```

```python
# child / derived class

class smartphone(Phone):
    def __init__(self, model, battery, camera, processor):
        super().__init__(model, battery, 48, processor)
        self.processor = processor

    def charge(self, hour):
        print("fast charging in process")
        super().charge(hour)

oneplus = smartphone("Oneplus Nord", 4190, 100, "Snapdragon")    

oneplus.charge(1)

print(oneplus.model)
print(oneplus.charge)


class Cooling_Mechanism:
    def __init__(self, cooling_method):
        self.cooling_method = cooling_method

    # methods 
    def cooling_on(self):
        print(f"the system is being cool by {self.cooling_method}")

class Smartphone_CoolingMode(smartphone, Cooling_Mechanism):
    def __init__(self, model, battery, camera, processor, cooling_method):
        smartphone.__init__(self, model, battery, camera, processor)
        Cooling_Mechanism.__init__(self, cooling_method)


pro_cooling = Smartphone_CoolingMode("Vivo", 5000, 100, "Mediatech", "Nitrogen")

print(pro_cooling.processor) #smartphone class
print(pro_cooling.camera) #phone class tekhe
print(pro_cooling.cooling_method) #cooling_machanism class tekhe
print(pro_cooling.cooling_on()) #cooling_machanism class er method
print(pro_cooling.charge(1)) #smartphone er modified charge jeta phone class tekhe inharitance kora 
```

```python
class Camera:
    def __init__(self, name):
        self.name = name

    def capture(self):
        print("A photo is captured")
```

```python
class Smart_Phone(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

    # method overridding
    def capture(self):
        print("Phone is captured by phone")

class DSLR(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

    # method overridding
    def capture(self):
        print("Phone is captured by dslr")

class Drone(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

       # method overridding
    def capture(self):
        print("Phone is captured by drone")

Phone = Smart_Phone("Phone", 30)
dslr = DSLR("DSLR", 200)
drone = Drone("Drone", 150)


Phone.capture()
dslr.capture()
drone.capture()
```
