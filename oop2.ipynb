```python
class Camera:
    def __init__(self, name):
        self.name = name

    def capture(self):
        print("A photo is captured")
```

```python
class Smart_Phone(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

    # method overridding
    def capture(self):
        print("Phone is captured by phone")

class DSLR(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

    # method overridding
    def capture(self):
        print("Phone is captured by dslr")

class Drone(Camera):
    def __init__(self, name, resulation):
        super().__init__(name)
        self.resulation = resulation

       # method overridding
    def capture(self):
        print("Phone is captured by drone")


Phone = Smart_Phone("Phone", 30)
dslr = DSLR("DSLR", 200)
drone = Drone("Drone", 150)


Phone.capture()
dslr.capture()
drone.capture()
```
